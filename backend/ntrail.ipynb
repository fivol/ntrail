## Данные 
*Все данные делятся на 4 типа*
### 1. Сырые данные. (full_data) Вся информация полученная с помощью соответствущего апи.
    - Содержат все необходимое
    - Очень много весят
    - Существует на уровне каждого класса информационного объекта. (VKUser, VKCommunity, IGUser, IGCommunity и пр)


### 2. Обработанные данные. (processed_data) Получены из сырых. 
    - Мало весят (выжимка из сырых данных)
    - Содержат личную информацию (ники людей в сообществе, телефоны, и прочие частные вещи)
    - Отсутствует маловажная информация (редко встечающаяся, шумовая, строковая)


### 3. Самое важное. Общая информация (clean_data).
    - Это выжимка из обработанных данных
    - Убраны все частности, только общие числовие или категориальные параметры
    - Предназначино для машинного обучения и алгоритмической приоретизации данных
    - Полностью сохраняется в базе данных для дальнейшего анализа. Обучения машинки


### 4. Пользовательская информация (output_data).
    - То, что выводится непосредственно потребителю на фронт.
    - Зависит от уровня доступа, привилегий, подписки и прочего.
    - Зависит от типа запроса (общий пробив, получение конкретики и др)
    - Формируется из всех 3х предыдущий пунктов:
        - Доступ к full_data только по целевому запросу пользователя
        - Результат работы модельки МЛ и алгоритмической обработки clean_data
        - Самое важное из processed_data на основе анализа clean_data
        - Получение исходников любого параметра

```python
def find_all_friends(target_user, assumptions=None, iterations=100):
    if not assumptions:
        assumptions = VKCommunity()
        
    assumptions += VKCommunity([target_user])
        
    possible_frinds = assumptions.friends()
    possible_frinds = possible_frinds.only_valid()
    print('possible_frinds', possible_frinds.size)
    possible_frinds.friends()
    possible_frinds.short_data
    friends = []
    i = 0
    used = set()
    while i < possible_frinds.size:
        print('Iteration:', i, '  Possible friends:', possible_frinds.size)
        if i % 50 == 0:
            best = possible_frinds.select(200)
            best.friends()
            best.short_data
        user = possible_frinds.select(1)[0]
        used.add(user)
        possible_frinds.counter[user.id] = 0
        if user.valid:
            if target_user in user.friends():
                user.print()
                friends += [user]
                fr = user.friends()
                fr = VKCommunity([node for node in fr.nodes if node not in used])
                possible_frinds = possible_frinds + fr
        i += 1
        if i > iterations:
            break
                
    return VKCommunity(friends)
```

```python
key_importance = []
target_community = comm1.get_params()

for key in params:
    target_element = target_community[key]
#     print(target_element)
    hist_list = param_list(key)
    bins = 10
    hist_values, hist_x = np.histogram(hist_list, bins)
#     plt.hist(hist_list, bins)
#     plt.show()
#     print(hist_x)
    if target_element < hist_x[0]:
        hist_value = 0
    elif target_element > hist_x[-1]:
        hist_value = 0
    else:
        hist_column = np.where(hist_x <= target_element)[0][-1]
        hist_value = hist_values[hist_column]

    hist_max = np.max(hist_values)
    element_importance = hist_value / hist_max
    key_importance.append((element_importance, key))
#     print(hist_max)
#     print(element_fr)
[print(i) for i in sorted(key_importance)]
None
```
